```python
import os
import sys
from pathlib import Path


# I don't use notebooks a lot so these are good checks for me to make sure the environment is set up correctly - Kostas
print("Python executable:", sys.executable)
print("Current working directory:", Path.cwd())
os.environ["OMPI_MCA_btl"] = "^uct"
from amuse.units import units

print("1 solar mass =", (1 | units.MSun).in_(units.kg))
```

```python
from amuse.datamodel import Particle
from amuse.community.seba import Seba
from amuse.units import units

print("Starting SeBa...", flush=True)

# If this runs for more than 30 seconds, something is wrong. Maybe check out the issue I have in the github - Kostas
stellar = Seba()

try:
    print("Setting metallicity...", flush=True)
    stellar.parameters.metallicity = 0.02

    print("Creating star...", flush=True)
    star = stellar.particles.add_particle(
        Particle(mass=1.2 | units.MSun)
    )

    print("Evolving star...", flush=True)
    stellar.evolve_model(1 | units.Myr)

    print("\nSuccess!")
    print("Age:", stellar.model_time.in_(units.Myr))
    print("Mass:", star.mass.in_(units.MSun))
    print("Radius:", star.radius.in_(units.RSun))
    print("Luminosity:", star.luminosity.in_(units.LSun))

finally:
    stellar.stop()
```
