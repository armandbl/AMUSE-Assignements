```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Locate the white_dwarf_project root (the folder that contains src/) and put it
# on sys.path so `from src.visualization import load_data` works no matter where
# the Jupyter kernel was started.
def _project_root():
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / "src" / "visualization.py").is_file():
            return candidate
    raise FileNotFoundError("Could not find the white_dwarf_project root (src/visualization.py).")

sys.path.insert(0, str(_project_root()))
datadir = _project_root() / "reference"
data = pd.read_csv(datadir / "planet_survival_results.csv", comment="#")


```

```python
# remove Model stop cases from the data, but keep them in a separate variable for later analysis
# data = data[data["case_result"] != "Model stop"]


data_20 = data[data["initial_a_out_AU"] == 20]
data_15 = data[data["initial_a_out_AU"] == 15]


# plt.figure(figsize=(10, 6))
# plt.scatter(data_20["a0_AU"], data_20["inclination_deg"], label="data", alpha=0.7, color='blue')
# plt.xlabel("Initial a_in (AU)")
# plt.ylabel("Final a_in (AU)")
# plt.title("Final vs Initial a_in for Different Initial a_out Values")
# plt.legend()
# plt.grid()
# plt.show()

not_finished = data[data["case_result"] == "Model stop"]
survived = data[data["case_result"] == "Reached WD endpoint"]
contact = data[data["case_result"] == "Surface contact"]
dissagree = data[data["case_result"] == "Settings dissagree"]

not_finished_20 = not_finished[not_finished["initial_a_out_AU"] == 20]
not_finished_15 = not_finished[not_finished["initial_a_out_AU"] == 15]

dissagree_20 = dissagree[dissagree["initial_a_out_AU"] == 20]
dissagree_15 = dissagree[dissagree["initial_a_out_AU"] == 15]

contact_20 = contact[contact["initial_a_out_AU"] == 20]
contact_15 = contact[contact["initial_a_out_AU"] == 15]

survived_20 = survived[survived["initial_a_out_AU"] == 20]
survived_15 = survived[survived["initial_a_out_AU"] == 15]

def plot_a_out(data, a_out_value):
    data_filtered = data[data["initial_a_out_AU"] == a_out_value]
    for case_result in data_filtered["case_result"].unique():
        case_data = data_filtered[data_filtered["case_result"] == case_result]
        plt.scatter(case_data["a0_AU"], case_data["inclination_deg"], label=case_result, alpha=0.7)
    plt.figure(figsize=(10, 6))
    plt.xlabel("Initial a_in (AU)")
    plt.ylabel("Final a_in (AU)")
    plt.title(f"Initial a_out = {a_out_value} AU")
    plt.legend(data_filtered["case_result"].unique())
    plt.grid()
    plt.show()

# plot_a_out(data, 20)
# plot_a_out(data, 15)

plt.figure(figsize=(10, 6))
plt.scatter(survived_20["a0_AU"], survived_20["inclination_deg"], label="Survived", alpha=0.7, color='blue')
plt.scatter(contact_20["a0_AU"], contact_20["inclination_deg"], label="Contact", alpha=0.7, color='red')
plt.scatter(dissagree_20["a0_AU"], dissagree_20["inclination_deg"], label="Dissagree", alpha=0.7, color='orange')
plt.scatter(not_finished_20["a0_AU"], not_finished_20["inclination_deg"], label="Not Finished", alpha=0.7, color='green')
plt.xlabel("Initial a_in (AU)")
plt.ylabel("Inclination (deg)")
plt.title("Initial a_in vs inclination for Initial a_out = 20 AU")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(survived_15["a0_AU"], survived_15["inclination_deg"], label="Survived", alpha=0.7, color='blue')
plt.scatter(contact_15["a0_AU"], contact_15["inclination_deg"], label="Contact", alpha=0.7, color='red')
plt.scatter(dissagree_15["a0_AU"], dissagree_15["inclination_deg"], label="Dissagree", alpha=0.7, color='orange')
plt.scatter(not_finished_15["a0_AU"], not_finished_15["inclination_deg"], label="Not Finished", alpha=0.7, color='green')
plt.xlabel("Initial a_in (AU)")
plt.ylabel("Final a_in (AU)")
plt.title("Final vs Initial a_in for Initial a_out = 15 AU")
plt.legend()
plt.grid()
plt.show()
```
